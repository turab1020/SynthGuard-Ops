# 01 · EDA — UCI Adult (`adult.data`)
Run `dvc repro download` first. Reads the raw file exactly as the pipeline does.

```python
import pandas as pd
from pathlib import Path
from synthguard.data.validate import read_raw
from synthguard.data.schema import NUMERIC, CATEGORICAL

root = Path.cwd() if (Path.cwd() / 'params.yaml').exists() else Path.cwd().parent
df = read_raw(root / 'data/raw/adult.data')
print(df.shape)
df.head()
```

## Dtypes

```python
df.dtypes
```

## Missing values (`?`) per column

```python
q = (df[CATEGORICAL] == '?').sum()
q[q > 0].to_frame('n_question_marks').assign(pct=lambda t: (100 * t.n_question_marks / len(df)).round(2))
```

## Class balance

```python
df['income'].value_counts(normalize=True).round(4)
```

## Duplicates

```python
print('exact duplicate rows:', int(df.duplicated().sum()))
```

## Numeric ranges

```python
df[NUMERIC].describe().T[['min', 'max', 'mean', 'std']].round(2)
```

## Category cardinality

```python
df[CATEGORICAL].nunique().sort_values(ascending=False)
```
